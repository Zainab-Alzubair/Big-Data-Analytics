```python
import requests
r = requests.get('https://www.gisma.com/about-us', auth=('user', 'pass'))
r.status_code

r.headers['content-type']

r.encoding

r.text
```

```python
from bs4 import BeautifulSoup
soup = BeautifulSoup(r.text, 'html.parser')

print(soup.prettify())
```
